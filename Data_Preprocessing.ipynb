```python
sheets = ["NORTH-CENTRAL NIGERIA", "NORTH-WEST NIGERIA", "NORTH-EAST NIGERIA"]
dfs = []

for sheet in sheets:
    df = pd.read_excel("/content/drive/MyDrive/IIT/2nd year/2nd semester/Data Science Project Life Cycle/DSPL_ICW/nigerian-acute-malnutrition-2026-april-may-september.xlsx",
                       sheet_name=sheet, header=1, skiprows=[0])

    # Clean column names
    df.columns = df.columns.str.replace('\n', ' ').str.strip()

    # Add region column
    df["Region"] = sheet.replace(" NIGERIA", "").title()

    # Remove the "Total" row
    df = df[df.iloc[:, 0] != "Total"]

    # Remove any completely empty rows
    df = df.dropna(how='all')

    # Remove rows where the LGA column is empty
    df = df.dropna(subset=[df.columns[1]])

    dfs.append(df)

combined = pd.concat(dfs, ignore_index=True)

# Check it looks right
print(f"Shape: {combined.shape}")
print(f"Columns: {combined.columns.tolist()}")
print(f"\nAny 'Total' rows remaining: {combined.iloc[:,0].str.contains('Total', na=False).any()}")
combined.head(10)
```

```python
combined.to_csv("Nigeria_malnutrition_cleaned.csv", index=False)
from google.colab import files
files.download("Nigeria_malnutrition_cleaned.csv")
```
